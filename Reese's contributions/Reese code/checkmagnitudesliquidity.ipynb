```python
import pandas as pd

# 1. Load the CSV
path = r"D:\vc-research\vc-research\Reese's contributions\Reese postprocessing outputs\company_date_total_liquidation.csv"
df = pd.read_csv(path)

# 2. Make sure filing_date is a datetime
df["filing_date"] = pd.to_datetime(df["filing_date"], errors="coerce")

# 3. Sort by company_id and date
df = df.sort_values(["company_id", "filing_date"])

# 4. Get previous total_liquidation_amount within each company_id
df["prev_total_liquidation_amount"] = (
    df.groupby("company_id")["total_liquidation_amount"].shift(1)
)

# 5. Create flag: 1 if previous > current, 0 otherwise
df["prev_higher_flag"] = (
    df["prev_total_liquidation_amount"] > df["total_liquidation_amount"]
).astype(int)

# (Optional) If you’d rather have the first observation be NaN instead of 0:
# df.loc[df["prev_total_liquidation_amount"].isna(), "prev_higher_flag"] = pd.NA

# 6. (Optional) Save back out
df.to_csv(r"D:\vc-research\vc-research\Reese's contributions\Reese postprocessing outputs\company_date_total_liquidation_with_valuationdecrease_flag.csv",           index=False)
```
